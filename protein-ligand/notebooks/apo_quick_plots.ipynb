```python
import os, re, numpy as np
import matplotlib.pyplot as plt

def read_xvg(path):
    meta = {"xaxis": None, "yaxis": None}
    rows = []
    with open(path, "r") as f:
        for line in f:
            s = line.strip()
            if not s or s.startswith("#"):
                continue
            if s.startswith("@"):
                m = re.search(r'@\s+xaxis\s+label\s+"(.+)"', s)
                if m: meta["xaxis"] = m.group(1)
                m = re.search(r'@\s+yaxis\s+label\s+"(.+)"', s)
                if m: meta["yaxis"] = m.group(1)
                continue
            try:
                rows.append([float(v) for v in s.split()])
            except:
                continue
    arr = np.array(rows, dtype=float)
    x = arr[:, 0]
    y = arr[:, 1:]   # keep all Y columns (multi-column files)
    if y.ndim == 1:
        y = y.reshape(-1, 1)
    return x, y, meta

def plot_single(x, y, title, xlabel, ylabel, out_png, label):
    plt.figure(figsize=(14, 5.2))
    plt.plot(x, y, linewidth=2.0, label=label)
    plt.title(title, fontsize=16)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True, alpha=0.25)
    plt.xlim(float(x.min()), float(x.max()))
    # legend outside on the right, aligned with the top
    plt.legend(frameon=False, loc="upper left", bbox_to_anchor=(1.02, 1.0))
    plt.tight_layout()
    plt.savefig(out_png, dpi=300)
    plt.close()

# ====== FILES (APO) ======
RMSD_XVG = "rmsd_apo_clean.xvg"
RMSF_XVG = "rmsf_apo_res.xvg"
RG_XVG   = "rg_apo_clean.xvg"
SASA_XVG = "sasa_protein_apo.xvg"

# ====== RMSD ======
x, y, _ = read_xvg(RMSD_XVG)
plot_single(x, y[:,0], "RMSD (APO)", "Time (ns)", "RMSD (nm)", "APO_RMSD.png", "BRAF (apo)")

# ====== RMSF (Cα) ======
x, y, m = read_xvg(RMSF_XVG)
plot_single(x, y[:,0], "RMSF (APO)", m["xaxis"] or "Residue", "RMSF (nm)", "APO_RMSF_CA.png", "BRAF (Cα)")

# ====== Rg ======
x, y, _ = read_xvg(RG_XVG)
plot_single(x, y[:,0], "Radius of gyration (Rg, APO)", "Time (ns)", "Rg (nm)", "APO_Rg.png", "BRAF")

# ====== SASA protein ======
x, y, m = read_xvg(SASA_XVG)
plot_single(x, y[:,0], "SASA (APO)", "Time (ns)", m["yaxis"] or "Area (nm$^2$)", "APO_SASA_Protein.png", "BRAF (APO)")

print("Done: APO_RMSD.png, APO_RMSF_CA.png, APO_Rg.png, APO_SASA_Protein.png")
```
